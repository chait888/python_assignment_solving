```python
#3.Simple Grader – Given a score (0‑100), print "A" for 90‑100, "B" for 80‑89, "C" for 70‑79, "D" for 60‑69, else "F"
```

```python
score=int(input("Enter the score between 0 to 100:"))
if 90<=score<=100:
    print("A")
elif 80<=score<=89:
    print("B")
elif 70<=score<=79:
    print("C")
elif 60<=score<=69:
    print("D")
else:
    print("F")
    
    
    
    
    
```

```python
  # 6.Count Vowels – Read a string and count how many vowels it contains (a, e, i, o, u; case‑insensitive) using a for loop
```

```python
s=str(input("Enter the string:"))
vowels="aeiouAEIOU"
count=0
for i in s:
    if i in vowels:
        count=count+1
count
```

# 10.List Min & Max (No built‑ins) – Given a list of integers, determine the minimum and maximum values using a single pass for loop without min() or max()

```python
l=[23,45,56,78,89,12,45]
minimum=l[0]
maximum=l[1]
for i in l:
   if i>minimum:
       minimum=i
   if i>maximum==i:
       maximum=i
print(minimum)
print(maximum)
```
