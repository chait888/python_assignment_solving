```python
   # Take a user’s age as input and determine the movie‑ticket price using if‑elif‑else logic:

#< 5 → “Free”, 5–18 → “₹50”, 18–60 → “₹100”, ≥ 60 → “₹70”
```

```python
age=int(input("Enter the age:"))
if age<=5:
    print("free")
elif 5<age<18:
    print("rupees50")
elif 18<age<60:
    print("rupees100")
else:
    print("rupees70")
    
```

```python
# Check if a given year is a leap year using the standard Gregorian rules
```

```python
year=int(input("Enter the year:"))
if (year%4==0 and year%100!=0) or year%400==0:
    print("Leap year")
else:
    print("Not leap year")
          
```
